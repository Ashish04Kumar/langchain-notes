```python
import random

class NakliLLm:
    def __init__(self):
        print('LLM Created')
    
    def predict(self, prompt):
        # dummy resp
        response_list=[
            "Delhi is the capital of India",
            "IPL is a cricket league",
            "A.I stands for artificial intelligence"
        ]
        return {'response': random.choice(response_list)}
```

```python
llm = NakliLLm()
llm.predict("What is the capital of India")
```

```python
# str = "Write a poem about {topic}"
# result = str.format(topic="nature")

class NakliPromptTemplate:
    def __init__(self, template, input_variables):
        self.template = template
        self.input_variables = input_variables

    def format(self, input_dict):

        # Check required variables
        for variable in self.input_variables:
            if variable not in input_dict:
                raise ValueError(f"Missing variable: {variable}")

        return self.template.format(**input_dict)

```

```python
promptTemplate = NakliPromptTemplate(
    template = "Write a {length} poem about {topic}",
    input_variables = ['length','topic']
)

prompt = promptTemplate.format({'length': 'short','topic': 'India'})
prompt
```

```python
llm = NakliLLm()
llm.predict(prompt)
```

```python
class NakliLLMChain:
    def __init__(self, llm, prompt):
        self.llm = llm
        self.prompt = prompt

    def run(self,input_dict):
        final_prompt = self.prompt.format(input_dict)
        result = self.llm.predict(final_prompt)
        return result['response']
```

```python
llm = NakliLLm()
promptTemplate = NakliPromptTemplate(
    template = "Write a {length} poem about {topic}",
    input_variables = ['length','topic']
)
```

```python
chain = NakliLLMChain(llm, promptTemplate)
chain.run({'length': 'short', 'topic' : 'India'})
```
